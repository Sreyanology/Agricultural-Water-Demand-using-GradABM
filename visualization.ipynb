```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
states = gpd.read_file('India Shape/india_st.shp')

highlight_states = ['PUNJAB', 'MAHARASHTRA', 'WEST BENGAL']
highlighted = states[states['STATE'].isin(highlight_states)]

fig, ax = plt.subplots(figsize=(10, 12))
states.plot(ax=ax, color='#e6e6e6', edgecolor='white', linewidth=0.8)
highlighted.plot(ax=ax, color='#d95f02', edgecolor='black', linewidth=1.2)

for _, row in highlighted.iterrows():
    point = row.geometry.representative_point()
    ax.text(point.x, point.y, row['STATE'].title(), fontsize=10, ha='center')

ax.set_title('India Map Highlighting Punjab, Maharashtra, and West Bengal', fontsize=14)
ax.set_axis_off()
plt.show()
```
